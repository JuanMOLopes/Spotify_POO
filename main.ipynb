## Teste de duplicidade

```python
try:
    from catalogo import Catalogo
    print('Classe Catalogo importada com sucesso!')
    
except ImportError as erro:
    print(f'Erro ao importar a classe Catalogo: {erro}')
```

```python
musicas = [m1, m2, m3, m4, m5, m6, m7]

c2 = Catalogo()

for musica in musicas:
    c2.adicionar_musica(musica)
print(f'Total de músicas adicionadas: {len(c2.musicas)} musicas')
```

```python
repetida = Musica('hate that i made you love me', 'Ariana Grande', 'Pop', 198, 2026)

print(f'Duplicada? {c2.verificar_duplicidade(repetida)}')
print(f'Cadastro realizado? {c2.adicionar_musica(repetida)}')
print(f'Total após a tentativa: {len(c2.musicas)} musicas')
```

```python
repetida = Musica('Hate That I Made You Love Me', 'ariana grande', 'POP', 198, 2026)

print(f'Duplicada? {c2.verificar_duplicidade(repetida)}')
print(f'Cadastro realizado? {c2.adicionar_musica(repetida)}')
print(f'Total após a tentativa: {len(c2.musicas)} musicas')
```

## Teste de filtros

```python
# Filtro por nome
print(f'Filtrar por nome')
for musica in c2.filtrar_por_nome("proibida"):
    print(musica)
    
print(f'\n{"="*100}')

# Filtro por cantor
print(f'\nFiltrar por cantor')
for musica in c2.filtrar_por_cantor('ariana'):
    print(musica)
    
print(f'\n{"="*100}')

# Filtro por gênero
print(f'\nFiltrar por gênero')
for musica in c2.filtrar_por_genero('rock'):
    print(musica)
    
print(f'\n{"="*100}')

# Filtro por duração
print(f'\nFiltrar por duração')
for musica in c2.filtrar_por_duracao(3, 5):
    print(musica)

print(f'\n{"="*100}')

# Filtro por ano
print(f'\nFiltrar por ano')
for musica in c2.filtrar_por_ano(2024):
    print(musica)
```
