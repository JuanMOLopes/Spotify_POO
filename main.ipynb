## Lista de músicas

```python
m2 = Musica('THE GREATEST', 'Billie Eilish', 'Pop', 293, 2024)
m3 = Musica('EVIL', 'Melanie Martinez', 'Pop', 246, 2023)
m4 = Musica('hate that i made you love me', 'Ariana Grande', 'Pop', 198, 2026)
m5 = Musica('Proibida Pra Mim (Grazon)', 'Charlie Brown Jr.', 'Rock Alternativo', 168, 1997)
m6 = Musica('Zóio de Lula', 'Charlie Brown Jr.', 'Rock Alternativo', 252, 1999)
m7 = Musica('Dias de Luta, Dias de Glória', 'Charlie Brown Jr.', 'Rock Alternativo', 145, 2005)

musicas = [m1, m2, m3, m4, m5, m6, m7]

for musica in musicas:
    print(musica)
```
